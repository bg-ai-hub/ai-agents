```python
judge_message = [{"role": "user", "content": judge}]
```

```python

openai = OpenAI()
response = openai.chat.completions.create(
    model="gpt-5-mini",
    messages=judge_message
)

results = response.choices[0].message.content
print("Judge's ranking of the competitors:")
print(results)
```

```python
results_dict = json.loads(results)
ranks = results_dict["results"]
for index,results in enumerate(ranks):
    competitor_index = int(results) - 1
    print(f"Rank {index+1}: Competitor {results} - {competitors[competitor_index]}")
```

## Debate/Discussion Pattern

### Round 1 - Initial Arguments

```python
question_prompt = "suggest a nuanced question in Corporate Team Management that I can ask a number of LLMs to evaluate they intelligence. Give me only question, " \
"no explanation"
message = [{"role": "user", "content": question_prompt}]
```

```python
openai = OpenAI()
response = openai.chat.completions.create(
    model="gpt-4o-mini",
    messages=message,
    temperature=0.7,
)
new_question = response.choices[0].message.content
print("New question generated by OpenAI's GPT-4o-mini:")
display(Markdown(f"**{new_question}**"))
```

```python
responses = {}
time_taken_dict = {}
llms = ["gpt-4o-mini", "gemini-2.5-flash", "deepseek-chat", "llama-3.2:latest", "groq-llama-3.3-70b-versatile"]

for llm in llms:
    start_time = time.time()
    print(f"Getting response from {llm}...")
    if llm == "gpt-4o-mini":
        response = openai.chat.completions.create(
            model=llm,
            messages=[{"role": "user", "content": new_question}]
        )

    elif llm == "gemini-2.5-flash":
        response = gemini.chat.completions.create(
            model=llm,
            messages=[{"role": "user", "content": new_question}]
        )

    elif llm == "deepseek-chat":
        response = deepseek.chat.completions.create(
            model=llm,
            messages=[{"role": "user", "content": new_question}]
        )

    elif llm == "llama-3.2:latest":
        ollama = OpenAI(
            base_url="http://localhost:11434/v1",api_key="ollama"
        )
        model_name = "llama3.2:latest"

        response = ollama.chat.completions.create(
            model=model_name,
            messages=messages
        )

    elif llm == "groq-llama-3.3-70b-versatile":
        groq = OpenAI(
            api_key=groq_api_key,
            base_url="https://api.groq.com/openai/v1"
        )

        model_name = "llama-3.3-70b-versatile"

        response = groq.chat.completions.create(
            model=model_name,
            messages=messages
        )
    end_time = time.time()
    time_taken = end_time - start_time
    answer = response.choices[0].message.content
    responses[llm] = answer
    time_taken_dict[llm] = f"{time_taken:.2f}"
    print(f"Time taken to get response from {llm}: {time_taken:.2f} seconds")
```

```python
for llm, answer in responses.items():
    print(f"{llm} response:\n{answer}\n")
```
